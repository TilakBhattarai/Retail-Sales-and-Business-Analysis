### Numerical Investigation

Objective: Understand the distribution, characteristics, and unusual values of important numerical columns.

#### 1. Sales Analysis

Objective: Investigate sales values, their range, average, variation, and unusual values.

1. Minimum and Maximum Sales

```python
minSales = dataset_df['Sales'].min()
print(f"The minimum sales is: {minSales}")
```

```python
maxSales = dataset_df['Sales'].max()
print(f"The maximum sales is: {maxSales}")
```

2. Mean and Median Sales

```python
mean = dataset_df['Sales'].mean()
print(f"The mean sales is: {mean:.2f}")
```

```python
median = dataset_df['Sales'].median()
print(f"The median sales is: {median:.2f}")
```

3. Variation in Sales Values

1. Standard Deviation

```python
sales_sd = dataset_df['Sales'].std()
print("Sales standard deviation:", sales_sd)
```

2. Interquartile Range (IQR)

```python
Q1 = dataset_df['Sales'].quantile(0.25)
Q3 = dataset_df['Sales'].quantile(0.75)

IQR =  Q3 - Q1
print("Sales IQR:", IQR)
```

4. Unusual or Negative Values

```python
sales = dataset_df['Sales']
negatives = dataset_df[sales < 0]
print("Negative values:", len(negatives))
```

```python
lower, upper =  Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
unusual = dataset_df[(sales < lower) | (sales > upper)]
print(f"Normal range: {lower:.2f} to {upper:.2f}")
print("Unusual values:", len(unusual))
```

Observation:

- Minimum sales: 0.444
- Maximum sales: 22,638.48
- Average sales: 228.23
- Median sales: 53.91
- Standard deviation: 619.91
- IQR: 192.28
- Negative sales values: None.
- Unusual sales values: 1,183 records identified using the IQR method.

The average is considerably higher than the median, indicating a right-skewed sales distribution. The standard deviation also shows considerable variation in sales values.

The IQR method identified 1,183 unusually high sales records above the upper threshold of 497.92. These values are statistical outliers and should be investigated further rather than automatically treated as errors.

Overall, sales values vary considerably, with a small number of high-value transactions influencing the average.

## Date Investigation

Investigating date columns to understand the dataset's time coverage and identify potential date inconsistencies.

### 1. Order Date Investigation

Objective: Identify the earliest and latest order dates.

1. Order Date Range

```python
dataset_df['Order Date'] = pd.to_datetime(dataset_df['Order Date'])

print("Order Date type:", dataset_df['Order Date'].dtype)
print("Earliest order date:", dataset_df['Order Date'].min())
print("Latest order date:", dataset_df['Order Date'].max())
```

Order dates range from January 3, 2023, to December 30, 2026. However, the latest date is in the future relative to the current date (November 14, 2026), so future-dated records should be investigated before time-based analysis.

### 2. Ship Date Investigation

Objective: Identify the earliest and latest shipping dates.

1. Ship Date Range

```python
print("Ship Date type:", dataset_df['Ship Date'].dtype)
dataset_df['Ship Date'] = pd.to_datetime(dataset_df['Ship Date'])

print("Earliest ship date:", dataset_df['Ship Date'].min())
print("Latest ship date:", dataset_df['Ship Date'].max())
```

Observations:

Ship dates range from January 7, 2023, to January 5, 2027. The latest shipping date is in the future relative to November 14, 2026, requiring further investigation to verify the validity of future-dated records.

### 3. Date Consistency Check

Objective: Identify invalid dates and check whether shipping dates are consistent with order dates.

1. Date Consistency Investigation

```python
print("Invalid order dates:", dataset_df['Order Date'].isnull().sum())
print("Invalid ship dates:",  dataset_df['Ship Date'].isnull().sum())

ship_days = (dataset_df['Ship Date'] - dataset_df['Order Date']).dt.days
invalid_ship = ship_days[ship_days < 0]
print("Ship date before order date:", len(invalid_ship))

print("Min shipping days:", ship_days.min())
print("Max shipping days:", ship_days.max())
print("Average shipping days:", round(ship_days.mean(), 2))
print("Most common shipping days:", ship_days.mode()[0])
print("Rows with that shipping time:", (ship_days == ship_days.mode()[0]).sum())
```

Observations:

No invalid dates or shipping date inconsistencies were found. Shipping duration ranges from 0 to 11 days, with an average of 3.96 days. The most common shipping duration is 4 days, occurring in 2,838 records. However, future-dated records identified in the previous investigation still require verification.
