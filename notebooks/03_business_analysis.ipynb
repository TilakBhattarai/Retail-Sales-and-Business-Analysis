## 2. Business Performance Analysis

Objective: Calculate the total sales, total profit, order volume, customer count, and quantity sold.

The overall profit margin is 12.56%, which means the company earns about 12.56 in profit for every 100 in sales. The average order value is 455.20, calculated per unique order, not per order line. These are overall figures across all four years, including the future-dated records retained in Day 2.

### 2.3 Business Performance Summary

Key Findings:

- Overall profit margin: 12.56%
- Average order value: 455.20

```python
print(sales_df.iloc[0:5, 0:3])
```

### 3.3 Boolean Filtering

Objective: Select rows that satisfy one or more conditions.


#### 3.3.1 Filter Loss-Making Records

Observations:

Filtering Profit Status == 'Loss' returned 1,901 records, which matches the Day 1 count of loss-making order-line records. Combining two conditions (Category == 'Technology' and Profit > 0) with & returned 1,591 profitable Technology records.

### 3.4 Sorting

Objective: Sort records to find the highest and lowest values.

#### 3.4.1 Top 5 Records by Sales

```python
sales_df.sort_values('Sales', ascending=False).head()[['Order ID', 'Order Date' ,'Product Name', 'Sales', "Profit"]]
```

#### 3.4.2 Bottom 5 Records by Profit

Key Findings:

- Single brackets (df['col']) select one column as a Series; double brackets (df[['col1', 'col2']]) select several columns as a DataFrame.
- iloc selects rows and columns by position and excludes the end index; loc selects by label and includes the end index.
- Boolean filtering keeps only rows that satisfy a condition; multiple conditions are combined with & and each condition is placed in brackets.
- sort_values orders rows by a column; ascending=False gives the largest values first.
- These operations return a new object and do not change the original dataset.

## 4. Sales & Profit Trends
Objective: Investigate sales and profit performance over time to identify important trends and patterns.

### 4.1 Yearly Performance

Objective: Calculate total sales and total profit for each year.

#### 4.1.1 Yearly Sales

```python
yearly_sales = (sales_df.groupby('Order Year')['Sales'].sum())
print("Yearly sales:")
print(yearly_sales)
```

#### 4.1.2 Yearly Profit

```python
yearly_profit = (sales_df.groupby('Order Year')['Profit'].sum())
print("Yearly profit:")
print(yearly_profit)
```

#### 4.1.3 Verify Yearly Totals and Profit Margin

```python
print("Yearly sales add up to total sales:", np.isclose(yearly_sales.sum(), total_sales))
print("Yearly profit add up to total profit:", np.isclose(yearly_profit.sum(), total_profit))

yearly_margin = yearly_profit / yearly_sales * 100
print(yearly_margin.round(2))

yearly_sales_growth = yearly_sales.pct_change() * 100

print("Year-over-year sales growth (%):")
print(yearly_sales_growth.round(2))
```

Observations:

- Yearly sales were 494,040.21 in 2023, 472,993.03 in 2024, 613,933.58 in 2025 and 745,567.53 in 2026. Sales dipped about 4.3% from 2023 to 2024, then grew about 29.8% in 2025 and about 21.4% in 2026.
- Yearly profit rose every year: 51,684.30 in 2023, 62,020.97 in 2024, 82,665.20 in 2025 and 95,926.35 in 2026.
- 2026 has the highest sales and profit, but it includes the 709 future-dated records (after November 15, 2026) retained in Day 2, so its comparison with earlier years should be treated with this caveat.
- The yearly totals add up to the overall total sales and total profit, confirming that no records were lost in the grouping.
- The yearly profit margin was 10.46% in 2023, 13.11% in 2024, 13.46% in 2025 and 12.87% in 2026. Profit grew faster than sales in 2024 and 2025, while in 2026 sales grew faster than profit, so the margin slipped slightly from its 2025 peak.

### 4.2 Monthly Performance

Objective: Calculate total sales and total profit for each month.

#### 4.2.1 Monthly Sales

```python
monthly_sales = sales_df.groupby('Year-Month')['Sales'].sum()
print("Monthly sales:")
print(monthly_sales)
```

#### 4.2.2 Monthly Profit

```python
monthly_profit = sales_df.groupby('Year-Month')['Profit'].sum()
print("Monthly Profit:")
print(monthly_profit)
```

Observations:

- Monthly sales and profit were calculated for each of the 48 months from 2023-01 to 2026-12.
- Sales are lowest at the start of the year (for example 2023-02 at 4,519.89 and 2024-02 at 11,951.41) and higher in September, November and December (for example 2023-09 at 82,670.43 and 2024-11 at 75,972.56), which suggests a seasonal pattern.
- Monthly profit is positive in most visible months but turned negative in 2023-07 (-841.48) and 2024-01 (-3,189.80), even though sales were positive in those months.
- The highest sales value visible in the output is 118,454.51 in 2026-11. The 2026-11 figure includes future-dated records after November 15, and all of 2026-12 consists of future-dated records, so both months should be read with this caveat.

### 4.3 Best and Worst Performing Periods

Objective: Identify the months with the highest and lowest sales and profit, and count loss-making months.

#### 4.3.1 Best and Worst Months by Sales

```python
best_sales_month = monthly_sales.idxmax()
worst_sales_month = monthly_sales.idxmin()

print(f"Best sales month: {best_sales_month} ({monthly_sales.max():,.2f})")
print(f"Worst sales month: {worst_sales_month} ({monthly_sales.min():,.2f})")
```

#### 4.3.2 Best and Worst Months by Profit

```python
best_profit_month = monthly_profit.idxmax()
worst_profit_month = monthly_profit.idxmin()

print(f"Best profit month: {best_profit_month} ({monthly_profit.max():,.2f})")
print(f"Worst profit month: {worst_profit_month} ({monthly_profit.min():,.2f})")
```

#### 4.3.3 Loss-Making Months

```python
loss_months = monthly_profit[monthly_profit < 0]
print("Loss making months:", len(loss_months))
print(loss_months)
```

Observations:

- The highest monthly sales were in 2026-11 (118,454.50) and the lowest in 2023-02 (4,519.89). The 2026-11 figure includes future-dated records after November 15, 2026, so it should be read with this caveat.
- The highest monthly profit was in 2025-12 (17,926.30) and the lowest in 2024-01 (-3,189.80).
- Only 2 of the 48 months were loss-making: 2023-07 (-841.48) and 2024-01 (-3,189.80). Both months still had positive sales, so the losses came from low profit on the sales made, not from missing sales.

### 4.4 Sales & Profit Trends — Summary

Key Findings:

- Yearly sales were 494,040.21 in 2023, 472,993.03 in 2024, 613,933.58 in 2025 and 745,567.53 in 2026. Sales dipped about 4.3% in 2024, then grew about 29.8% in 2025 and about 21.4% in 2026.
- Yearly profit rose every year, from 51,684.30 in 2023 to 95,926.35 in 2026.
- The yearly profit margin was 10.46% in 2023, 13.11% in 2024, 13.46% in 2025 and 12.87% in 2026.
- Sales are lowest early in the year and higher in September, November and December, which suggests a seasonal pattern.
- The best sales month was 2026-11 (118,454.50) and the best profit month was 2025-12 (17,926.30). The worst sales month was 2023-02 (4,519.89).
- Only 2 of 48 months were loss-making: 2023-07 and 2024-01.

Note: 2026 figures include the 709 future-dated records retained in Day 2, so 2026 and its November and December months are not directly comparable with earlier years.

Decision / Conclusion:

Sales and profit grew over the four years, and profit grew even in 2024 when sales dipped, which means the margin improved. Most months were profitable, with seasonal peaks late in the year. The causes behind the 2024 dip and the two loss-making months are not explained by this data alone and are left for later analysis.

## 5. Product Performance Analysis

Objective: Understand product performance by examining sales and profitability across different product classifications.

### 5.1 Category Performance

Objective: Calculate total sales, total profit, and profit margin for each product category.

#### 5.1.1 Sales by Category

```python
category_sales = sales_df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
print("Sales by category:")
print(category_sales)
```

#### 5.1.2 Profit by Category

```python
category_profit = sales_df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
print("Profit by category:")
print(category_profit)
```

#### 5.1.3 Profit Margin by Category

```python
# 5.1.3 Profit Margin by Category
category_margin = (category_profit / category_sales * 100).round(2)
print("Profit margin by category (%):")
print(category_margin)

print("\nCategory sales add up to total sales:", np.isclose(category_sales.sum(), total_sales))
print("Category profit adds up to total profit:", np.isclose(category_profit.sum(), total_profit))
```

Observations:

- Technology has the highest sales at 839,893.28, followed by Furniture at 754,747.76 and Office Supplies at 731,893.31. The three categories add up to the overall total sales.
- Technology also has the highest profit at 146,543.38, followed by Office Supplies at 126,023.44 and Furniture at 19,729.99. The three categories add up to the overall total profit.
- Profit margins are 17.45% for Technology, 17.22% for Office Supplies and 2.61% for Furniture.
- Furniture has the second-highest sales but the lowest profit, because its margin is far below the other two categories. Office Supplies has the lowest sales but the second-highest profit.

### 5.2 Sub-Category Performance

Objective: Calculate total sales, total profit, and profit margin for each sub-category.

#### 5.2.1 Sales by Sub-Category

```python
subcategory_sales = sales_df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
print("Sales by sub-category:")
print(subcategory_sales)
```

#### 5.2.2 Profit by Sub-Category

```python
subcategory_profit = sales_df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)
print("Profit by sub-category:")
print(subcategory_profit)
```

#### 5.2.3 Profit Margin by Sub-Category

```python
subcategory_margin = (subcategory_profit / subcategory_sales * 100).round(2)
print("Profit margin by sub-category (%):")
print(subcategory_margin)


print("\nSub-Category sales add up to total sales:", np.isclose(subcategory_sales.sum(), total_sales))
print("Sub-category profit adds up to total profit:", np.isclose(subcategory_profit.sum(), total_profit))
```

Observations:

- Chairs (335,768.25), Phones (331,842.64) and Storage (224,644.55) have the highest sales among the 17 sub-categories.
- Copiers (56,093.94), Phones (45,050.83) and Accessories (41,936.64) have the highest profit. Copiers has only the eighth-highest sales (150,745.29) but the highest profit.
- Three sub-categories are loss-making: Tables (-17,753.21), Bookcases (-3,632.07) and Supplies (-1,171.39). Tables has the fourth-highest sales (208,020.18) but the largest loss, with a margin of -8.53%.
- Labels (43.90%), Paper (43.39%) and Envelopes (42.28%) have the highest profit margins, while Tables (-8.53%), Bookcases (-3.15%) and Supplies (-2.51%) have negative margins. Machines has sales of 189,925.03 but a margin of only 1.82%.
- Tables and Bookcases belong to the Furniture category, so they explain why Furniture has a low margin (2.61%) despite high sales.
- The sub-category totals add up to the overall total sales and total profit.
- Within Furniture, Tables and Bookcases are loss-making, which pulls down the category's profit (see 5.2).

### 5.3 High and Low Performing Product Groups

Objective: Identify the best and worst sub-categories by profit and by profit margin.

#### 5.3.1 Top and Bottom Sub-Categories by Profit

```python
top_profit_sub = subcategory_profit.head(3)
bottom_profit_sub = subcategory_profit.tail(3).sort_values(ascending=True)

print("\nTop 3 sub-categories by profit:\n")
print(top_profit_sub)

print("\nBottom 3 sub-categories by profit:\n")
print(bottom_profit_sub)
```

#### 5.3.2 Top and Bottom Sub-Categories by Profit Margin

```python
top_margin_sub = subcategory_margin.sort_values(ascending=False).head(3)
bottom_margin_sub = subcategory_margin.sort_values(ascending=False).tail(3)

print("\nTop 3 sub-categories by profit margin:\n")
print(top_margin_sub)

print("\nBottom 3 sub-categories by profit margin:\n")
print(bottom_margin_sub)
```

Observations:

- Copiers, Phones and Accessories are the top three sub-categories by total profit.
- Tables, Bookcases and Supplies are the bottom three sub-categories by total profit, and all three are loss-making.
- Labels (43.90%), Paper (43.39%) and Envelopes (42.28%) have the highest profit margins.
- Supplies (-2.51%), Bookcases (-3.15%) and Tables (-8.53%) have the lowest profit margins.
- Copiers generate the highest total profit, while Labels have the highest profit margin, showing that the most profitable sub-category by total amount is not necessarily the most profitable relative to sales.
